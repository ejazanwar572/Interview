Problem Description:

You’re given two dataframes: `transactions` and `products`.  
  
The transactions dataframe contains transaction ids, product ids, and the total amount of each
product sold.

The products dataframe contains product ids and prices.

Write a function to return a dataframe containing every transaction with a total value of over $100.
Include the total value of the transaction as a new column in the dataframe.

**Example:**

**Input:**

`df_transactions`

**transaction_id** | **product_id** | **amount**  
---|---|---  
1 | 101 | 3  
2 | 102 | 5  
3 | 103 | 8  
4 | 104 | 3  
5 | 105 | 2  

`df_products`

**product_id** | **price**  
---|---  
101 | 20.00  
102 | 21.00  
103 | 15.00  
104 | 16.00  
105 | 52.00

**Output:**

transaction_id | product_id | amount | total_value  
---|---|---|---  
2 | 102 | 5 | 105.00  
3 | 103 | 8 | 120.00  
5 | 105 | 2 | 104.00


```python
# Question 24: Over 100 Dollars
# 
# You’re given two dataframes: transactions and products.
#
# The transactions dataframe contains transaction ids, product ids, and the total amount
# of each
# product sold.
#
# The products dataframe contains product ids and prices.
#
# Write a function to return a dataframe containing every transaction with a total value
# of over $100.
# Include the total value of the transaction as a new column in the dataframe.

import pandas as pd

# Mock Data
products_data = {
    'product_id': [1, 2, 3, 4],
    'price': [10.50, 25.00, 5.00, 55.00]
}
products = pd.DataFrame(products_data)

transactions_data = {
    'transaction_id': [101, 101, 102, 103, 103, 104],
    'product_id': [1, 2, 4, 1, 3, 2],
    'amount': [5, 2, 3, 15, 10, 1] # quantity sold
}
transactions = pd.DataFrame(transactions_data)

def solution(transactions_df, products_df):
    df = pd.merge(products_df, transactions_df, on='product_id')
    df['total_value'] = df['amount'] * df['price']
    
    # Group by transaction_id and sum both the 'amount' (quantity) and 'total_value'
    # df = df.groupby(['transaction_id','product_id','amount'])['total_value'].sum().reset_index()
    
    # Filter the aggregated dataframe
    df = df[df['total_value'] > 100]
    
    return df

solution(transactions, products)
```
