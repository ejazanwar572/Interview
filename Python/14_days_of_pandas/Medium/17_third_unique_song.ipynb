Problem Description:
===================
Given a table of `song_plays` and a table of `users`, write a query to extract the earliest date
each user played their third unique song.  
  
_Note: If a user has listened to less than three unique songs, display their name but with a`NULL`
date and song name_

**Example:**

**Input:**

`song_plays` table

Columns | Type  
---|---  
`user_id` | INTEGER  
`song_name` | TEXT  
`date_played` | DATETIME  
  
**users** table

Columns | Type  
---|---  
`id` | INTEGER  
`name` | VARCHAR  
  
**Output:**

Columns | Type  
---|---  
`name` | VARCHAR  
`date_played` | DATETIME  
`song_name` | TEXT


```python
# Question 30: Third Unique Song
# 
# Find the third unique song each user listened to.

import pandas as pd

# Mock Data (15 listens, 3 users)
listens_data = {
    'id': list(range(1, 16)),
    'user_id': [
        1, 1, 1, 1, 1,    # User 1: listens to 4 unique songs (repeated some)
        2, 2, 2, 2,       # User 2: listens to 3 unique songs
        3, 3, 3, 3, 3, 3, # User 3: listens to only 2 unique songs (no 3rd!) ❌
    ],
    'song_name': [
        'SongA', 'SongB', 'SongA', 'SongC', 'SongD',  # User 1: 3rd unique = SongC
        'SongX', 'SongY', 'SongX', 'SongZ',            # User 2: 3rd unique = SongZ
        'SongM', 'SongN', 'SongM', 'SongN', 'SongM', 'SongN',  # User 3: only 2 unique ❌
    ],
    'listen_date': pd.to_datetime([
        '2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05',
        '2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04',
        '2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05', '2024-01-06',
    ])
}
listens = pd.DataFrame(listens_data)
# Expected: User 1 → SongC, User 2 → SongZ, User 3 → NULL/excluded

def solution():
    # Write your solution here
    df = listens
    df = df.sort_values(by=['user_id','listen_date'])
    df = df.drop_duplicates(subset=['user_id','song_name'])
    df['song_rank'] = df.groupby('user_id')['listen_date'].rank(method='first').astype(int)
    df = df[df['song_rank']==3]
    return df

solution()
```
