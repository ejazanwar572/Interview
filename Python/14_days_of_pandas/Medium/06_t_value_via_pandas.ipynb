Question 14: t Value via Pandas
Difficulty: Medium
Link: https://www.interviewquery.com/questions/t-value-via-pandas?playlist=14-days-of-pandas

Problem Description:
===================
You are given a dataframe with a single column, `'var'`.  
  
Calculated the t-value for the mean of ‘`var`’ against a null hypothesis that \\(μ = μ_0\\).

_Note: You do not have to calculate the p-value of the test or run the test._

**Example:**

**Input:**

    
    
    mu0 = 1
    print(df)
    ...
       var
    0  -34
    1   40
    2  -89
    3    5
    4  -26
    

**Output:**

    
    
    def t_score(mu0, df) ->
    var   -1.015614
    dtype: float64
