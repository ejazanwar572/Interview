```python
# Altertnate sols

def solution(df):
    # Count the events using boolean masking (faster & more idiomatic)
    chat_start = (df['event_type'] == 'audio_chat_start').sum()
    chat_end = (df['event_type'] == 'audio_chat_end').sum()

    # Edge case: prevent ZeroDivisionError if no chats started
    if chat_start == 0:
        return 0.0

    # Calculate and round to 2 decimal places
    succ_rate = (chat_end * 100) / chat_start
    return round(succ_rate, 2)

# Call the function by passing the dataframe
solution(events)
```
