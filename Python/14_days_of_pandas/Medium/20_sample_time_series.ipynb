Problem Description:
===================
Given a transactions table with date timestamps, sample every 4th row ordered by the date.

**Example:**

**Input:**

`transactions` table

Columns | Type  
---|---  
`id` | INTEGER  
`user_id` | INTEGER  
`created_at` | DATETIME  
`product_id` | INTEGER  
`quantity` | INTEGER  
  
**Output:**

Columns | Type  
---|---  
`created_at` | DATETIME  
`product_id` | INTEGER


```python
# Question 33: Sample Time Series
# 
# Write code to resample or process a time series dataset.

import pandas as pd
import numpy as np

# Mock Data — 30 days of daily data with realistic patterns
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=30, freq='D')
ts_data = {
    'date': dates,
    'value': np.round(np.random.normal(100, 15, 30), 2),  # Normal dist around 100
    'category': ['A', 'B', 'C'] * 10,  # Rotating categories
}
# Inject edge cases
ts_data['value'][5] = np.nan   # Missing value on Jan 6
ts_data['value'][15] = np.nan  # Missing value on Jan 16
ts_data['value'][25] = 500.0   # Outlier spike on Jan 26
ts_data['value'][0] = 0.0      # Zero on first day

df_ts = pd.DataFrame(ts_data)

def solution():
    # Write your solution here
    df = df_ts.sort_values(by = 'date')
    df = df.iloc[::2,0]
    return df

solution()
```

```python
# Optimal Solution Review:
# 1. The problem asks for every *4th* row, so the step should be `::4` rather than `::2`.
# 2. It is best practice to pass the dataframe as a parameter rather than relying on global variables.
# 3. `iloc[::4, :]` returns all columns. `iloc[::4, 0]` only returns the first column as a Series.

def optimal_solution(df):
    # Sort chronologically
    sorted_df = df.sort_values(by='date')
    
    # Sample every 4th row
    sampled_df = sorted_df.iloc[::4, :]
    
    return sampled_df

optimal_solution(df_ts)
```
