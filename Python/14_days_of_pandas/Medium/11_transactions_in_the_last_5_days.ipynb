Problem Description:
===================
Let’s say you work at a bank.  
  
Using the bank_transactions table, find how many users made at least one transaction each day in the
first five days of January 2020.

`bank_transactions` table

Column | Type  
---|---  
`user_id` | INTEGER  
`created_at` | DATETIME  
`transaction_value` | FLOAT  
`id` | INTEGER  
  
**Output:**

Column | Type  
---|---  
`number_of_users` | INTEGER


```python
# Question 22: Transactions in the Last 5 Days
# 
# Let's say you work at a bank.
# Using the bank_transactions table, find how many users made at least one transaction
# each day in the first five days of January 2020.

import pandas as pd

# Mock Data
transactions_data = {
    'id': list(range(1, 21)),
    'user_id': [
        1, 1, 1, 1, 1, 1,   # User 1: transacts ALL 5 days (+ extra on Jan 1) ✅
        2, 2, 2, 2, 2,       # User 2: transacts ALL 5 days ✅
        3, 3, 3,             # User 3: only Jan 1, 2, 3 — misses day 4 & 5 ❌
        4, 4,                # User 4: only Jan 1, 5 — misses days 2, 3, 4 ❌
        5, 5,                # User 5: Jan 6 & Jan 10 — outside window ❌
        6, 6,                # User 6: Dec 31 & Jan 1 — partial ❌
    ],
    'created_at': pd.to_datetime([
        '2020-01-01', '2020-01-01', '2020-01-02', '2020-01-03', '2020-01-04', '2020-01-05',  # User 1
        '2020-01-01', '2020-01-02', '2020-01-03', '2020-01-04', '2020-01-05',                # User 2
        '2020-01-01', '2020-01-02', '2020-01-03',                                            # User 3
        '2020-01-01', '2020-01-05',                                                          # User 4
        '2020-01-06', '2020-01-10',                                                          # User 5
        '2019-12-31', '2020-01-01',                                                          # User 6
    ]),
    'transaction_value': [
        10.5, 5.0, 20.0, 15.0, 8.0, 12.0,
        25.0, 30.0, 10.0, 45.0, 60.0,
        100.0, 200.0, 50.0,
        75.0, 90.0,
        15.0, 25.0,
        40.0, 55.0,
    ]
}
bank_transactions = pd.DataFrame(transactions_data)
# Expected answer: 2 (only User 1 and User 2 transacted all 5 days)

df = bank_transactions.copy() 

df = df[df['created_at'].between('2020-01-01','2020-01-05')].groupby(['user_id','created_at']).size().reset_index()


df = df.sort_values(by=['user_id','created_at'])

df['rank'] = df.groupby('user_id')['created_at'].rank(method = 'first').astype(int)

df['streak'] = df['created_at'] - pd.to_timedelta(df['rank'],unit='D')

df = df.groupby('user_id')['streak'].size().reset_index()

df[df['streak'] >= 5]

# df = df.groupby('user_id')['created_at'].rank(method='dense')
```

```python
# ✅ Optimal Solution

df = bank_transactions.copy()

# 1. Filter to first 5 days of Jan 2020
df = df[df['created_at'].between('2020-01-01', '2020-01-05')]

# 2. Count distinct days per user (normalise date in case of timestamps)
days_per_user = df.groupby('user_id')['created_at'].nunique()

# days_per_user

# 3. Count users who transacted on ALL 5 days
number_of_users = (days_per_user == 5).sum()

number_of_users

# pd.DataFrame({'number_of_users': [number_of_users]})
```
