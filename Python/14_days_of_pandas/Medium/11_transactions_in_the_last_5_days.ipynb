Question 22: Transactions in the Last 5 Days
Difficulty: Medium
Link: https://www.interviewquery.com/questions/transactions-in-the-last-5-days?playlist=14-days-of-pandas

Problem Description:
===================
Let’s say you work at a bank.  
  
Using the bank_transactions table, find how many users made at least one transaction each day in the
first five days of January 2020.

`bank_transactions` table

Column | Type  
---|---  
`user_id` | INTEGER  
`created_at` | DATETIME  
`transaction_value` | FLOAT  
`id` | INTEGER  
  
**Output:**

Column | Type  
---|---  
`number_of_users` | INTEGER
