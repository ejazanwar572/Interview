### Problem Description
Normalize a column using Min-Max scaling.

### Sample Input and Output
**Input:**
Column `revenue` with broad magnitude.

**Output:**
Column `revenue_normalized` bound between 0 and 1.

```
import pandas as pd

df = pd.DataFrame({'revenue': [100, 500, 1000]})
```

### Approach
Apply transformation: `(x - min) / (max - min)`. Good for models sensitive to scale.

```
df['revenue_normalized'] = (df['revenue'] - df['revenue'].min()) / (df['revenue'].max() - df['revenue'].min())
print(df)
```
