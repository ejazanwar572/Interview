Handling Large Files in Python
===
Difficulty: Medium

Problem Description:
===================
How do you handle a file that is too large to fit into memory in Python? Show an example.

Sample Input:
A large dataset CSV.

Sample Output:
Processed data (e.g., aggregated results or top N rows) without MemoryError.

```python
import pandas as pd
import os
import csv

# Mock Data Setup: Let's create a dummy large-ish file for demonstration
with open('large_sample.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['id', 'value'])
    for i in range(100):  # Mocking a large file
        writer.writerow([i, i*10])
```

**Concepts to use:**
1. **Chunking:** Process the file in smaller parts (e.g., using `pandas.read_csv(chunksize=...)`).
2. **Generators:** Yielding one row at a time instead of loading all at once.
3. **Dask or PySpark:** For distributed or parallel out-of-core computation.

```python
# Optimized Solution using Pandas chunksize

def process_large_file(filepath):
    chunk_size = 10  # Very small chunk for demo purposes
    total_sum = 0
    
    # Read file in chunks to avoid MemoryError
    for chunk in pd.read_csv(filepath, chunksize=chunk_size):
        # Process each chunk
        total_sum += chunk['value'].sum()
        
    return total_sum

result = process_large_file('large_sample.csv')
print(f"Total sum: {result}")

# Cleanup
os.remove('large_sample.csv')
```

Mutable vs Immutable Data Types
===
Difficulty: Easy

Problem Description:
===================
What is the difference between mutable and immutable data types? Give examples.

Sample Input:
`[1, 2, 3]` vs `(1, 2, 3)`

Sample Output:
Explanation and verification via `id()`.

```python
# Mock Data / Setup
mutable_example = [1, 2, 3]
immutable_example = (1, 2, 3)
```

**Concepts to use:**
1. **Mutable:** State or contents CAN be changed after creation (e.g., list, dict, set).
2. **Immutable:** State or contents CANNOT be changed after creation (e.g., int, float, str, tuple, frozenset).
3. **Memory/ID:** When you "modify" an immutable object, a new object is created in memory (checked via `id()`).

```python
# Optimized Explanation & Code Example
def demonstrate_mutability():
    print("--- Mutable Example (List) ---")
    my_list = [1, 2, 3]
    list_id = id(my_list)
    print(f"Original List: {my_list}, ID: {list_id}")
    
    my_list.append(4)
    print(f"Modified List: {my_list}, ID: {id(my_list)} (Same ID? {list_id == id(my_list)})")
    
    print("\n--- Immutable Example (Tuple & String) ---")
    my_str = "Hello"
    str_id = id(my_str)
    print(f"Original String: '{my_str}', ID: {str_id}")
    
    # We cannot modify the string inplace, we must reassign
    my_str += " World"
    print(f"Modified String: '{my_str}', ID: {id(my_str)} (Same ID? {str_id == id(my_str)})")

demonstrate_mutability()
```

Merge Multiple CSV Files
===
Difficulty: Easy

Problem Description:
===================
Write Python code to merge multiple CSV files from a directory into a single Pandas DataFrame.

Sample Input:
file1.csv and file2.csv in a directory.

Sample Output:
A single concatenated pandas DataFrame containing rows from both.

```python
import pandas as pd
import os

# Setup Mock Data: Create two dummy CSV documents
os.makedirs('dummy_csvs', exist_ok=True)

df1 = pd.DataFrame({'id': [1, 2], 'name': ['Alice', 'Bob']})
df2 = pd.DataFrame({'id': [3, 4], 'name': ['Charlie', 'Dave']})

df1.to_csv('dummy_csvs/file1.csv', index=False)
df2.to_csv('dummy_csvs/file2.csv', index=False)
```

**Concepts to use:**
1. **File discovery:** `glob.glob` or `os.listdir` to find all `.csv` files.
2. **Iteration:** A list comprehension to read each file into a DataFrame using `pd.read_csv()`.
3. **Concatenation:** `pd.concat()` to efficiently join the list of DataFrames into one.

```python
# Optimized Solution
import glob

def merge_csvs_in_directory(directory_path):
    # Find all CSV files in the directory
    all_files = glob.glob(os.path.join(directory_path, "*.csv"))
    
    if not all_files:
        return pd.DataFrame()
        
    # Read each CSV and store DataFrames in a list
    df_list = [pd.read_csv(file) for file in all_files]
    
    # Concatenate all DataFrames
    merged_df = pd.concat(df_list, ignore_index=True)
    return merged_df

# Test it on our mock data
final_df = merge_csvs_in_directory('dummy_csvs')
display(final_df)

# Cleanup
for f in glob.glob("dummy_csvs/*.csv"):
    os.remove(f)
os.rmdir('dummy_csvs')
```
