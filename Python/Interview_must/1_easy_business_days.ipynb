Business Days Between Two Dates
===
Difficulty: Easy

Problem Description:
===================
Given two dates `start` and `end`, return the number of **business days** (Mon–Fri, excluding weekends)
between them (inclusive of start, exclusive of end — same convention as Python ranges).

Sample Input:
```
start = '2024-01-01'   # Monday
end   = '2024-01-08'   # Monday (next week)
```

Sample Output:
```
Business days: 5   (Mon Jan 1 → Fri Jan 5)
```

````python
import numpy as np
import pandas as pd

# Sample Data
start = '2024-01-01'   # Monday
end   = '2024-01-08'   # Next Monday
````

**Concepts to use:**
1. **`np.busday_count(start, end)`** — counts business days between two dates; excludes weekends by default.
2. **`pd.bdate_range(start, end)`** — generates a DatetimeIndex of business days (alternative approach, gives you the actual dates).
3. **Manual approach:** `pd.date_range()` + filter `dayofweek < 5` (Mon=0, Sun=6).

````python
# Optimised Solution — Approach 1: numpy busday_count (fastest, 1 line)
def business_days_numpy(start, end):
    return int(np.busday_count(start, end))

print(f"numpy:       {business_days_numpy(start, end)} business days")

# Approach 2: pandas bdate_range (gives you the actual business dates)
def business_days_pandas(start, end):
    bdays = pd.bdate_range(start=start, end=end, inclusive='left')
    return len(bdays)

print(f"bdate_range: {business_days_pandas(start, end)} business days")

# Approach 3: manual filter (interview-friendly, shows understanding)
def business_days_manual(start, end):
    all_days = pd.date_range(start=start, end=end, inclusive='left')
    return (all_days.dayofweek < 5).sum()   # Mon=0 ... Fri=4

print(f"manual:      {business_days_manual(start, end)} business days")
````
