### Problem Description
Rotate a wide-format dataset to a long-format using `.melt()`, and back using `.pivot_table()`.

### Sample Input and Output
**Input:**
Wide format DataFrame (columns: jan_rev, feb_rev).

**Output:**
Long format (column: month, revenue), and reconstructed wide format.

```python
import pandas as pd

df_wide = pd.DataFrame({'user_id': [1, 2], 'jan_rev': [100, 200], 'feb_rev': [150, 250]})
```

### Approach
Using `melt()` to push wide columns into rows. `pivot_table` to pull rows horizontally out.

```python
# Wide to Long
df_long = df_wide.melt(id_vars=['user_id'], value_vars=['jan_rev', 'feb_rev'], var_name='month', value_name='revenue')
print(df_long)

# Long to Wide
df_reconstructed = df_long.pivot_table(index='user_id', columns='month', values='revenue', aggfunc='sum').reset_index()
print(df_reconstructed)
```
