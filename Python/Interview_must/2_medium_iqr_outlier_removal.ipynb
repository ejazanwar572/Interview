### Problem Description
Filter standard outliers using the IQR statistical method.

### Sample Input and Output
**Input:**
DataFrame with heavy outliers.

**Output:**
Clean DataFrame stripped of anomalies.

```
import pandas as pd

df = pd.DataFrame({'revenue': [10, 12, 14, 15, 1000, 11]})
```

### Approach
Find Q1 and Q3, compute the IQR range. Filter values lying outside [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR].

```
Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_clean = df[(df['revenue'] >= lower) & (df['revenue'] <= upper)]
print(df_clean)
```
