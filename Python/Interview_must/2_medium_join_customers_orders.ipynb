### Problem Description
Merge orders and customers DataFrames using a left join to preserve all orders.

### Sample Input and Output
**Input:**
`df_orders` and `df_customers`.

**Output:**
Enriched `df_orders`.

```
import pandas as pd

df_orders = pd.DataFrame({'order_id': [1, 2], 'customer_id': [101, 102]})
df_customers = pd.DataFrame({'customer_id': [101], 'name': ['Alice']})
```

### Approach
Default to left join when appending reference data onto transaction data. Review shape before and after to check multi-matches.

```
merged = pd.merge(df_orders, df_customers, on='customer_id', how='left')
print(merged)
```
