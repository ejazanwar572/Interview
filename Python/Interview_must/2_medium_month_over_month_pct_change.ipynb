Month-Over-Month Revenue % Change
===
Difficulty: Medium

Problem Description:
===================
Given a `transactions` table, calculate the **month-over-month percentage change** in total revenue.
Return each month with its total revenue and % change from the previous month.

Sample Input:
```
| order_id | amount | created_at |
|----------|--------|------------|
| 1        | 100    | 2021-01-05 |
| 2        | 200    | 2021-01-20 |
| 3        | 150    | 2021-02-10 |
| 4        | 300    | 2021-03-15 |
```

Sample Output:
```
| month   | revenue | pct_change |
|---------|---------|------------|
| 2021-01 | 300     | NaN        |
| 2021-02 | 150     | -50.0%     |
| 2021-03 | 300     | +100.0%    |
```

````python
import pandas as pd

transactions = pd.DataFrame({
    'order_id':   [1, 2, 3, 4, 5, 6, 7, 8],
    'amount':     [100, 200, 150, 80, 300, 220, 50, 400],
    'created_at': pd.to_datetime([
        '2021-01-05', '2021-01-20',
        '2021-02-10', '2021-02-22',
        '2021-03-15', '2021-03-28',
        '2021-04-01', '2021-04-18'
    ])
})
print(transactions)
````

**Concepts to use:**
1. **`dt.to_period('M')`** — converts datetime to a monthly period (e.g., `2021-01`).
2. **`groupby().sum()`** — aggregate revenue per month.
3. **`pct_change()`** — computes `(current - previous) / previous` for consecutive rows — the key method here.
4. **`round(2) * 100`** — convert fraction to readable percentage.

````python
# Optimised Solution
def month_over_month(df):
    df = df.copy()

    # Step 1: extract year-month period
    df['month'] = df['created_at'].dt.to_period('M')

    # Step 2: total revenue per month
    monthly = df.groupby('month')['amount'].sum().reset_index(name='revenue')

    # Step 3: MoM % change (pct_change gives fraction → multiply by 100)
    monthly['pct_change'] = monthly['revenue'].pct_change().mul(100).round(2)

    return monthly

print(month_over_month(transactions))
````
