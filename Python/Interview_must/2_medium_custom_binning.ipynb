### Problem Description
Transform a continuous column into categorized text bins dynamically.

### Sample Input and Output
**Input:**
`revenue` continuous numbers.

**Output:**
Categorized list (`low`, `medium`, `high`).

```
import pandas as pd

df = pd.DataFrame({'revenue': [50, 500, 5000]})
```

### Approach
Using `.apply()` running a custom python function row by row. (Note: use `pd.cut` or `np.select` for performance on larger datasets).

```
def revenue_tier(rev):
    if rev < 100:
        return 'low'
    elif rev < 1000:
        return 'medium'
    else:
        return 'high'

df['tier'] = df['revenue'].apply(revenue_tier)
print(df)
```
