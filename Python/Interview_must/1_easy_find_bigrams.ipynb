Find Bigrams from a String
===
Difficulty: Easy

Problem Description:
===================
Write a function `find_bigrams(sentence)` that takes a sentence and returns a list of all
consecutive two-word pairs (bigrams), sorted alphabetically.

Sample Input:
```
sentence = "the cat sat on the mat"
```

Sample Output:
```
[('cat', 'sat'), ('mat', None)... ]
# sorted: [('cat','sat'), ('on','the'), ('sat','on'), ('the','cat'), ('the','mat')]
```

More precisely: sorted list of tuples `(word[i], word[i+1])` for i in 0..n-2.

````python
# Sample Data
sentence = "the cat sat on the mat"
````

**Concepts to use:**
1. **`str.split()`** — tokenise the sentence into words.
2. **`zip(words, words[1:])`** — pair each word with the next one in a single pass — classic Python idiom.
3. **Sorting tuples** — `sorted()` on tuples sorts lexicographically by first element, then second.

````python
# Optimised Solution
def find_bigrams(sentence):
    words = sentence.split()
    bigrams = list(zip(words, words[1:]))  # zip pairs adjacent words
    return sorted(bigrams)

print(find_bigrams(sentence))

# Why zip(words, words[1:]) works:
# words      = ['the', 'cat', 'sat', 'on', 'the', 'mat']
# words[1:]  = ['cat', 'sat', 'on',  'the','mat']
# zip pairs  → ('the','cat'), ('cat','sat'), ... — stops at shorter list
````
