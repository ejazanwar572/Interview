Authors Viewing Own Articles
Source: LeetCode 1148
Difficulty: Easy

Problem Description:
===================
Table: `Views`

Column | Type  
---|---  
`article_id` | INTEGER  
`author_id` | INTEGER  
`viewer_id` | INTEGER  
`view_date` | DATE  
  
There is no primary key for this table, the table may have duplicate rows.  
Each row indicates that some viewer viewed an article (written by some author) on some date.  
Note that equal `author_id` and `viewer_id` indicate the same person.

**Task:**  
Find all the authors that viewed at least one of their own articles.  
Return the result table sorted by `id` in ascending order.

**Output:**

Column | Type  
---|---  
`id` | INTEGER


```python
# Authors Viewing Own Articles
# 
# Find all authors that viewed at least one of their own articles.
# Return result sorted by id ascending.

import pandas as pd

# Mock Data (12 rows with edge cases)
views_data = {
    'article_id': [1, 1, 1, 2, 2, 3, 3, 4, 5, 5, 6, 6],
    'author_id':  [3, 3, 2, 7, 7, 4, 4, 5, 8, 8, 9, 10],
    'viewer_id':  [
        3,   # Author 3 viewed own article ✅
        3,   # Duplicate row — same author viewing again (dedup test) ✅
        5,   # Author 2's article viewed by someone else ❌
        7,   # Author 7 viewed own article ✅
        1,   # Author 7's article viewed by someone else ❌
        4,   # Author 4 viewed own article ✅
        6,   # Author 4's article viewed by someone else ❌
        5,   # Author 5 viewed own article ✅
        2,   # Author 8's article viewed by someone else ❌
        6,   # Author 8's article viewed by someone else ❌
        10,  # Author 9's article viewed by author 10 ❌
        10,  # Author 10 viewed own article ✅
    ],
    'view_date': pd.to_datetime([
        '2024-01-01', '2024-01-01', '2024-01-02',
        '2024-01-01', '2024-01-03',
        '2024-01-02', '2024-01-02',
        '2024-01-03',
        '2024-01-01', '2024-01-04',
        '2024-01-05', '2024-01-05',
    ])
}
views = pd.DataFrame(views_data)
# Expected output: [3, 4, 5, 7, 10] — sorted ascending
# Author 2, 8, 9 should NOT appear (never viewed own article)
# Author 3 appears only ONCE despite 2 duplicate self-views

def solution():
    # Write your solution here
    df = views.copy()
    df = df[df['author_id'] == df['viewer_id']]
    result = df['author_id'].drop_duplicates().sort_values()
    return pd.DataFrame({'id':result.values})

solution()
```
