Interpolate Missing Time-Series Values
===
Difficulty: Medium

Problem Description:
===================
A climate sensor records daily temperatures but some readings are missing. Fill the missing values
using **linear interpolation** — estimate the missing value as the straight line between the
surrounding known values.

Sample Input:
```
| date       | temperature |
|------------|-------------|
| 2021-01-01 | 10.0        |
| 2021-01-02 | NaN         |
| 2021-01-03 | NaN         |
| 2021-01-04 | 16.0        |
```

Sample Output:
```
| date       | temperature |
|------------|-------------|
| 2021-01-01 | 10.0        |
| 2021-01-02 | 12.0        |  ← (10 + 16) / 3 steps
| 2021-01-03 | 14.0        |
| 2021-01-04 | 16.0        |
```

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'date':        pd.date_range('2021-01-01', periods=7),
    'temperature': [10.0, np.nan, np.nan, 16.0, np.nan, 22.0, 24.0]
})
print(df)
````

**Concepts to use:**
1. **`df.interpolate(method='linear')`** — fills NaNs by linear interpolation between neighbours.
2. **`method='time'`** — interpolates proportionally to actual time gaps (use when dates aren't evenly spaced).
3. **`limit` param** — `interpolate(limit=1)` fills at most 1 consecutive NaN (useful to avoid over-extrapolation).
4. **vs `ffill`** — ffill repeats last known value; interpolation estimates a midpoint — better for smooth signals.

````python
# Optimised Solution
def interpolate_temperatures(df):
    df = df.set_index('date').sort_index()

    # Linear interpolation (assumes equal time spacing)
    df['temperature_linear'] = df['temperature'].interpolate(method='linear')

    # Time-based interpolation (handles uneven gaps correctly)
    df['temperature_time']   = df['temperature'].interpolate(method='time')

    return df

print(interpolate_temperatures(df))
````
