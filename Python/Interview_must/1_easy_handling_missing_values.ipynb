### Problem Description
Demonstrate handling missing values in a Pandas DataFrame (dropping and filling).

### Sample Input and Output
**Input:**
DataFrame with NaNs.

**Output:**
Cleaned DataFrame.

```
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
```

### Approach
Inspect NaNs based on specific requirements to decide whether to drop rows `dropna()` or impute with `fillna()`.

```
# Drop rows with any missing values
df_dropped = df.dropna()

# Fill with the column mean
df_filled = df.fillna(df.mean())

print(df_dropped)
print(df_filled)
```
