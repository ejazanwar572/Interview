Average Ride Duration with Timedelta
===
Difficulty: Easy

Problem Description:
===================
You work at Uber. Given a DataFrame of rides with `started_at` and `ended_at` timestamps,
calculate the **average ride duration in minutes** for each city.

Sample Input:
```
| ride_id | city       | started_at           | ended_at             |
|---------|------------|----------------------|----------------------|
| 1       | New York   | 2021-01-01 08:00:00  | 2021-01-01 08:30:00  |
| 2       | New York   | 2021-01-01 09:00:00  | 2021-01-01 09:45:00  |
| 3       | London     | 2021-01-01 10:00:00  | 2021-01-01 10:20:00  |
```

Sample Output:
```
| city     | avg_duration_minutes |
|----------|----------------------|
| London   | 20.0                 |
| New York | 37.5                 |
```

````python
import pandas as pd

rides = pd.DataFrame({
    'ride_id':    [1, 2, 3, 4, 5],
    'city':       ['New York', 'New York', 'London', 'London', 'Chicago'],
    'started_at': pd.to_datetime(['2021-01-01 08:00', '2021-01-01 09:00',
                                  '2021-01-01 10:00', '2021-01-01 11:00',
                                  '2021-01-01 12:00']),
    'ended_at':   pd.to_datetime(['2021-01-01 08:30', '2021-01-01 09:45',
                                  '2021-01-01 10:20', '2021-01-01 11:50',
                                  '2021-01-01 12:15']),
})
print(rides)
````

**Concepts to use:**
1. **Timedelta subtraction** — subtracting two datetime columns gives a `timedelta` Series.
2. **`.dt.total_seconds()`** — converts timedelta to a float of seconds; divide by 60 for minutes.
3. **`groupby().mean()`** — average per group after converting to numeric.

````python
# Optimised Solution
def avg_ride_duration(df):
    df = df.copy()
    # Step 1: compute duration as total minutes (float)
    df['duration_min'] = (df['ended_at'] - df['started_at']).dt.total_seconds() / 60

    # Step 2: average per city
    result = (
        df.groupby('city')['duration_min']
          .mean()
          .reset_index()
          .rename(columns={'duration_min': 'avg_duration_minutes'})
          .sort_values('city')
    )
    return result

print(avg_ride_duration(rides))
````
