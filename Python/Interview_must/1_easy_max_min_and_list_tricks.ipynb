Find Maximum and Minimum in a List (Without Built-ins)
===
Difficulty: Easy

Problem Description:
===================
Given a list of integers, find both the maximum and minimum values **without** using Python's built-in `max()` or `min()` functions.

Sample Input:
```
a = [1, 26, 78, 54, 65]
```

Sample Output:
```
Max: 78
Min: 1
```

````python
# Sample Data
list = [1, 26, 78, 54, 65]
````

````python

max_val = min_val = list[0] 

for num in list[1:]:
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num
    
print('Max_value :',max_val)
print('Max_value :',min_val)     
    

    
````

**Concepts to use:**
1. **Initialisation:** Set both `max_val` and `min_val` to the first element `a[0]`.
2. **Linear scan:** Iterate through every element and compare with running max/min.
3. **Time complexity:** O(n) — single pass.  **Space complexity:** O(1).

````python
# Optimised Solution
def find_max_min(lst):
    if not lst:
        raise ValueError("List is empty")
    max_val = min_val = lst[0]
    for num in lst[1:]:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num
    return max_val, min_val

max_val, min_val = find_max_min(a)
print(f"Max: {max_val}")
print(f"Min: {min_val}")
````

Find Second Largest Without Sorting
===
Difficulty: Easy

Problem Description:
===================
Given a list of integers, find the **second largest** value without using `sorted()` or `sort()`.

Sample Input:
```
a = [10, 3, 45, 7, 45, 22]
```

Sample Output:
```
Second largest: 22
```

````python
# Sample Data
list = [10, 3, 45, 7, 45, 22]
````

````python
largest = list[0]
sec_large = list[0]

for num in list:
    if num > largest:
        largest = num
    elif num > sec_large and num != largest:
        sec_large = num

print('largest' , largest)
print('largest' , sec_large)
````

**Concepts to use:**
1. **Two-variable tracking:** Maintain `first` and `second` across one pass.
2. **Duplicate handling:** Skip updating `second` when element equals `first` to get *strictly* second largest.
3. **Edge case:** Handle lists with fewer than 2 distinct values (return `None`).

````python
# Optimised Solution
def second_largest(lst):
    if len(lst) < 2:
        raise ValueError("Need at least 2 elements")
    first = second = float('-inf')
    for num in lst:
        if num > first:
            second = first
            first = num
        elif num > second and num != first:  # strictly second
            second = num
    if second == float('-inf'):
        return None  # all elements are identical
    return second

print(f"Second largest: {second_largest(a)}")
````

Find Duplicates and Unique Elements in a List
===
Difficulty: Easy

Problem Description:
===================
Given a list of integers, find:
1. All **duplicate** values (appear more than once).
2. All **unique** values (appear exactly once).

Sample Input:
```
nums = [4, 1, 2, 9, 2, 4, 7, 1, 5]
```

Sample Output:
```
Duplicates : [1, 2, 4]
Unique     : [5, 7, 9]
```

````python
# Sample Data
nums = [4, 1, 2, 9, 2, 4, 7, 1, 5]
````

````python
dups = []
uniq = []

for item in nums:
    if item not in uniq:
        uniq.append(item)
    elif item in uniq:
        dups.append(item)

print(sorted(dups))
print(sorted(uniq))


````

**Concepts to use:**
1. **Frequency dict:** Use `dict.get()` or `collections.Counter` to count occurrences in O(n).
2. **Filter by count:** values with `count > 1` → duplicates; `count == 1` → unique.
3. **Quick existence check:** `len(lst) != len(set(lst))` tells you if *any* duplicates exist — O(n).
4. **Time complexity:** O(n).  **Space complexity:** O(n).

````python
# --- Approach 1: Manual frequency dict (interview-friendly) ---
def find_dups_and_unique(lst):
    freq = {}
    for num in lst:
        freq[num] = freq.get(num, 0) + 1

    duplicates = sorted([k for k, v in freq.items() if v > 1])
    unique     = sorted([k for k, v in freq.items() if v == 1])
    return duplicates, unique

dups, uniq = find_dups_and_unique(nums)
print(f"Duplicates : {dups}")
print(f"Unique     : {uniq}")

# --- Approach 2: collections.Counter (Pythonic, concise) ---
from collections import Counter
counts = Counter(nums)
print(f"\nCounter approach:")
print(f"Duplicates : {sorted(k for k, v in counts.items() if v > 1)}")
print(f"Unique     : {sorted(k for k, v in counts.items() if v == 1)}")

# --- Quick check: does the list have ANY duplicates at all? ---
has_dups = len(nums) != len(set(nums))
print(f"\nHas any duplicates: {has_dups}")
````
