Repeated Category Purchase
===
Difficulty: Easy

Problem Description:
===================
Given a DataFrame of user purchases, find all users who have purchased from the **same product
category more than once**. Return their `user_id` and `category`, sorted by `user_id`.

Sample Input:
```
| user_id | category    |
|---------|-------------|
| 1       | Electronics |
| 1       | Electronics |  ← repeated!
| 2       | Clothing    |
| 3       | Books       |
| 3       | Books       |  ← repeated!
| 4       | Electronics |
```

Sample Output:
```
| user_id | category    |
|---------|-------------|
| 1       | Electronics |
| 3       | Books       |
```

````python
import pandas as pd

purchases = pd.DataFrame({
    'user_id':  [1, 1, 2, 3, 3, 4, 1, 2],
    'category': ['Electronics', 'Electronics', 'Clothing', 'Books',
                 'Books', 'Electronics', 'Books', 'Clothing']
})
print(purchases)
````

**Concepts to use:**
1. **`groupby().size()`** — count rows per (user_id, category) pair.
2. **Boolean mask on grouped result** — filter where count > 1.
3. **`reset_index()`** — converts the GroupBy result back to a regular DataFrame.
4. Alternative: **`groupby().filter()`** — keeps all rows of groups satisfying a condition.

````python
# Optimised Solution — Approach 1: groupby + size filter
def repeated_category_purchase(df):
    counts = df.groupby(['user_id', 'category']).size().reset_index(name='count')
    result = counts[counts['count'] > 1][['user_id', 'category']]
    return result.sort_values('user_id').reset_index(drop=True)

print(repeated_category_purchase(purchases))

# Approach 2: groupby().filter() — returns all original rows matching condition
def repeated_category_purchase_v2(df):
    return (
        df.groupby(['user_id', 'category'])
          .filter(lambda g: len(g) > 1)
          .drop_duplicates()
          .sort_values('user_id')
          .reset_index(drop=True)
    )

print(repeated_category_purchase_v2(purchases))
````
