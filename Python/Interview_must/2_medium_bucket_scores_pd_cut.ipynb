Bucket Test Scores with pd.cut
===
Difficulty: Medium

Problem Description:
===================
Given a DataFrame of student test scores, assign each student a **grade bucket** using custom
score ranges: A (90-100), B (75-89), C (50-74), F (0-49). Return the count and cumulative
percentage per bucket.

Sample Input:
```
| student_id | test_score |
|------------|------------|
| 1          | 95         |
| 2          | 82         |
| 3          | 67         |
| 4          | 45         |
| 5          | 73         |
```

Sample Output:
```
| grade | count | cumulative_pct |
|-------|-------|----------------|
| F     | 1     | 20%            |
| C     | 2     | 60%            |
| B     | 1     | 80%            |
| A     | 1     | 100%           |
```

````python
import pandas as pd

df_grades = pd.DataFrame({
    'student_id': range(1, 13),
    'test_score': [100, 95, 90, 85, 80, 75, 70, 65, 50, 45, 30, 0]
})
print(df_grades)
````

**Concepts to use:**
1. **`pd.cut(series, bins, labels)`** — discretizes continuous values into bins; `right=True` means bins are `(low, high]`.
2. **`bins` list** — define the boundaries: `[0, 50, 75, 90, 100]` creates 4 buckets.
3. **`value_counts(sort=False)`** — count per category while preserving label order.
4. **`cumsum() / total * 100`** — running cumulative percentage.

````python
# Optimised Solution
def bucket_test_scores(df):
    bins   = [0, 50, 75, 90, 101]              # 101 to make 100 inclusive
    labels = ['F', 'C', 'B', 'A']

    df = df.copy()
    df['grade'] = pd.cut(
        df['test_score'],
        bins=bins,
        labels=labels,
        right=False,    # [low, high) — left-inclusive
        include_lowest=True
    )

    summary = (
        df['grade'].value_counts(sort=False)
                   .reset_index()
    )
    summary.columns = ['grade', 'count']
    summary = summary.sort_values('grade')     # F → C → B → A

    total = summary['count'].sum()
    summary['cumulative_pct'] = (summary['count'].cumsum() / total * 100).round(1)

    return summary

print(bucket_test_scores(df_grades))
````
