Impute Median by Group
===
Difficulty: Easy

Problem Description:
===================
You're given a dataframe of cheese prices from different regions. Some prices are missing (NaN).
Fill each missing price with the **median price of that region** (group-aware imputation).

Sample Input:
```
| cheese     | region | price |
|------------|--------|-------|
| Brie       | France | 12.0  |
| Camembert  | France | NaN   |
| Gouda      | Dutch  | 8.0   |
| Edam       | Dutch  | NaN   |
| Gouda      | Dutch  | 10.0  |
```

Sample Output:
```
| cheese     | region | price |
|------------|--------|-------|
| Brie       | France | 12.0  |
| Camembert  | France | 12.0  |  ← filled with France median
| Gouda      | Dutch  | 8.0   |
| Edam       | Dutch  | 9.0   |  ← filled with Dutch median (8+10)/2
| Gouda      | Dutch  | 10.0  |
```

````python
import pandas as pd
import numpy as np

df_cheeses = pd.DataFrame({
    'cheese':  ['Brie', 'Camembert', 'Gouda', 'Edam', 'Gouda', 'Cheddar', 'Stilton'],
    'region':  ['France', 'France', 'Dutch', 'Dutch', 'Dutch', 'UK', 'UK'],
    'price':   [12.0, np.nan, 8.0, np.nan, 10.0, 15.0, np.nan]
})
print(df_cheeses)
````

**Concepts to use:**
1. **`groupby().transform('median')`** — computes the group median and broadcasts it back to each row (same shape as original).
2. **`fillna()`** — fills only the NaN positions.
3. Key difference from simple `fillna(df['price'].median())` — that uses the *global* median, not the group median.

````python
# Optimised Solution
def impute_median_by_group(df, group_col, value_col):
    group_medians = df.groupby(group_col)[value_col].transform('median')
    df[value_col] = df[value_col].fillna(group_medians)
    return df

result = impute_median_by_group(df_cheeses.copy(), 'region', 'price')
print(result)
````
