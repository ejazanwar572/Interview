# 2. Advanced Merging
**Interview Gold:** Joins are where data bugs hide. Know your join types and validation.

### 2.1 Join Types & Suffixes
- **Inner**: Intersection of keys.
- **Outer**: Union of keys.
- **Left/Right**: Keys from one side.
- **Cross**: Cartesian product (careful with size!).

```python
users = pd.DataFrame({'id': [1, 2, 3, 4], 'name': ['Alice', 'Bob', 'Charlie', 'Dave']})
orders = pd.DataFrame({'id': [101, 102, 103, 104], 'user_id': [1, 2, 2, 5], 'amount': [100, 200, 150, 300]})

# --- Practice ---
# 1. Inner Join (Only users who ordered)
merged_inner = pd.merge(users, orders, left_on='id', right_on='user_id', suffixes=('_user', '_order'))

# 2. Left Join (All users, regardless of order)
merged_left = pd.merge(users, orders, left_on='id', right_on='user_id', how='left')

# 3. Outer Join (All users and all orders - see Dave and user_id 5)
merged_outer = pd.merge(users, orders, left_on='id', right_on='user_id', how='outer', indicator=True)

print("--- Outer Join with Indicator ---")
print(merged_outer)
```

### 2.2 Validation (`validate`) parameter
**Critical Feature:** Prevents duplicates from exploding your data volume unexpectedly.
- `'1:1'`: Checks if merge keys are unique in both left and right datasets.
- `'1:m'`: Checks if merge keys are unique in left dataset.
- `'m:1'`: Checks if merge keys are unique in right dataset.

**Scenario:** Merging `Transactions` (m) to `Users` (1).

```python
# --- Practice ---
# Try to merge users and orders with validate='1:1'. It should fail because user 2 has multiple orders.
try:
    pd.merge(users, orders, left_on='id', right_on='user_id', validate='1:1')
except Exception as e:
    print(f"Merge Failed as expected: {e}")

# Now try with correct validation '1:m'
pd.merge(users, orders, left_on='id', right_on='user_id', validate='1:m')
```

```python
df_sales = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 120, 90, 110],
    'Quantity': [10, 15, 20, 12, 9, 11]
})


# --- Practice
# 1. Dictionary-based aggregation: Returns original column names as headers.
# If multiple functions are applied to one column, it creates a MultiIndex.
dict_res = df_sales.groupby('Region').agg({'Sales': 'sum', 'Quantity': 'mean'})
print("Dictionary Aggregation (Original Names):\n", dict_res)

# 2. Named Aggregation: Allows custom column names and ensures a flat column index.
# This is the modern, recommended approach for clarity and downstream processing.
agg_df = df_sales.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Avg_Quantity=('Quantity', 'mean'),
    Max_Sale=('Sales', 'max')
)

print("\n--- Named Aggregation ---")
print(agg_df)
```

### 3.3 Grouped Window Functions
Essential for Time Series and Finance.
- `shift()`: Get previous row's value.
- `diff()`: Create a delta column.
- `pct_change()`: Return % growth.
- `cumsum()`: Running total.


```python
df_finance = pd.DataFrame({
    'Ticker': ['AAPL', 'AAPL', 'AAPL', 'GOOG', 'GOOG', 'GOOG'],
    'Day': [1, 2, 3, 1, 2, 3],
    'Price': [150, 152, 155, 200, 205, 202]
})

# --- Practice ---
# 1. Calculate daily price change
df_finance['Prev_Price'] = df_finance.groupby('Ticker')['Price'].shift(1)
df_finance['Daily_Change'] = df_finance.groupby('Ticker')['Price'].diff()
df_finance['Pct_Change'] = df_finance.groupby('Ticker')['Price'].pct_change() * 100

# 2. Cumulative Max Price seen so far
df_finance['Running_Max'] = df_finance.groupby('Ticker')['Price'].cummax()

print(df_finance)
```

# 3.5 Mastering GroupBy: Beyond the Basics
**The Split-Apply-Combine Pattern:**
1. **Split**: Break data into groups based on criteria.
2. **Apply**: Apply a function to each group independently.
3. **Combine**: Combine results into a data structure.

**Crucial Concept:** The `groupby()` object is *lazy*. It doesn't calculate anything until you ask it to.

```python
# Setup Data for GroupBy Deep Dive
df_gb = pd.DataFrame({
    'Team': ['A', 'A', 'A', 'B', 'B', 'C'],
    'Role': ['Dev', 'Dev', 'QA', 'Dev', 'QA', 'Dev'],
    'Salary': [100, 120, 80, 110, 90, 130],
    'Experience': [1, 3, 2, 2, 1, 5]
})

# --- 1. Aggregation Patterns ---
# Named Aggregation (The Cleanest Way)
agg_df = df_gb.groupby('Team').agg(
    Min_Salary=('Salary', 'min'),
    Max_Salary=('Salary', 'max'),
    Avg_Exp=('Experience', 'mean'),
    Team_Size=('Salary', 'count')  # Count of any non-null column
)

# Multiple functions on specific columns (The Dictionary Way)
dict_agg = df_gb.groupby('Team').agg({
    'Salary': ['min', 'max', 'mean'],
    'Experience': 'max'
})

print("--- Named Aggregation ---")
print(agg_df)
print("\n--- Dictionary Aggregation ---")
print(dict_agg)
```

### 2. The Power of Transform (`.transform`)
Unlike `agg` (which reduces rows), `transform` returns a sequence of the **same length** as the original group.
Use Cases:
- **Standardization:** (x - mean) / std
- **Imputation:** Fill missing values with the group mean.

```python
# --- 2. Transform Patterns ---

# Scenario: Find employees who earn more than their Team's average
df_gb['Team_Avg_Salary'] = df_gb.groupby('Team')['Salary'].transform('mean')
df_gb['Is_High_Earner'] = df_gb['Salary'] > df_gb['Team_Avg_Salary']

print("--- Transform Example (Team Avg) ---")
print(df_gb)

# Scenario: Z-Score (How many std devs away from the team mean?)
df_gb['Salary_Z_Score'] = df_gb.groupby('Team')['Salary'].transform(
    lambda x: (x - x.mean()) / x.std()
)
```

# 4. Vectorization & Operations
**Performance is key.** Loops in Python are slow. Vectorized operations in C (via numpy/pandas) are fast.

- **Avoid** `df.iterrows()` unless absolutely necessary.
- **Avoid** `df.apply()` for simple math (it's just a loop under the hood).
- **Use** `numpy.where` or `numpy.select` for conditional logic.

```python
# --- Practice ---
df_perf = pd.DataFrame({'val': np.random.randint(1, 100, 100000)})

# Scenario: if val > 50, 'High', else 'Low'

# 1. The Slow Way (Apply)
%timeit df_perf['val'].apply(lambda x: 'High' if x > 50 else 'Low')

# 2. The Fast Way (Vectorized / np.where)
%timeit np.where(df_perf['val'] > 50, 'High', 'Low')
```

### 7.3 Rolling Windows
- `rolling(window=n).mean()`: Moving Average.
- `expanding().mean()`: Average of all data up to current point.


```python
dates = pd.date_range('2023-01-01', periods=7)
df_ts = pd.DataFrame({'val': [1, 2, 3, 4, 10, 2, 3]}, index=dates)

# --- Practice ---
# 3-day Moving Average
df_ts['MA_3'] = df_ts['val'].rolling(window=3).mean()

# Expanding Mean (Cumulative Average)
df_ts['Cum_Avg'] = df_ts['val'].expanding().mean()

print(df_ts)
```

# 10. Capstone Project
**Objective:** Process raw sales data to calculate **Monthly Cohort Retention**.
*This is a classic difficult interview question.*

**Steps:**
1. Create datasets.
2. Determine the user's "first purchase month" (CohortMonth).
3. Join CohortMonth back to the transaction data.
4. Calculate the discrepancy between TransactionMonth and CohortMonth (CohortIndex).
5. Pivot to see retention count per cohort.

```python
# --- Data Setup ---
np.random.seed(42)
n_rows = 1000
dates = pd.date_range('2023-01-01', '2023-12-31')
users = np.random.randint(1, 200, n_rows)

df_capstone = pd.DataFrame({
    'TransactionDate': np.random.choice(dates, n_rows),
    'UserID': users,
    'amount': np.random.uniform(10, 100, n_rows)
})

# --- Solution Steps ---
# 1. Convert to month period (yyyy-mm)
df_capstone['OrderMonth'] = df_capstone['TransactionDate'].dt.to_period('M')

# 2. Find first purchase month for each user (Cohort Month)
df_capstone['CohortMonth'] = df_capstone.groupby('UserID')['TransactionDate'].transform('min').dt.to_period('M')

# 3. Calculate Cohort Index (Months since first purchase)
# We need to convert period arithmetic to integers
def get_month_diff(x, y):
    return (x.dt.year - y.dt.year) * 12 + x.dt.month - y.dt.month

# Using to_timestamp to extract year/month easily for math, or just period math
df_capstone['CohortIndex'] = (df_capstone['OrderMonth'] - df_capstone['CohortMonth']).apply(lambda x: x.n)

# 4. Group by CohortMonth and CohortIndex
cohort_data = df_capstone.groupby(['CohortMonth', 'CohortIndex'])['UserID'].nunique().reset_index()

# 5. Pivot Table
cohort_counts = cohort_data.pivot(index='CohortMonth', columns='CohortIndex', values='UserID')

# 6. Calculate Retention %
cohort_sizes = cohort_counts.iloc[:, 0] # Month 0 size
retention = cohort_counts.divide(cohort_sizes, axis=0)

# Display (Heatmap style logic)
print(retention.round(3) * 100)
```
