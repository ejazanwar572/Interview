# Practice 03: Python ETL - Data Quality & Aggregation

## Problem Description
Wells Fargo receives daily transaction logs from an external vendor. The data is messy. You must drop duplicate records, standardize customer ID casing, handle missing transaction amounts by imputing a value of 0, and return the total transaction amount per customer correctly.


## Sample Input and Output

**Input Table (`raw_logs`)**:
| cust_id | amount | date       |
|---------|--------|------------|
| c-123   | 50.0   | 2024-03-01 |
| C-123   | NaN    | 2024-03-02 |
| C-999   | 100.0  | 2024-03-01 |
| c-123   | 50.0   | 2024-03-01 | (Duplicate)

**Output Table**:
| cust_id | total_amount |
|---------|--------------|
| C-123   | 50.0         |
| C-999   | 100.0        |


```python
import pandas as pd
import numpy as np

# DDL / Data Frame setup
data = {
    'cust_id': ['c-123', 'C-123', 'C-999', 'c-123', 'C-444'],
    'amount': [50.0, np.nan, 100.0, 50.0, 200.0],
    'date': ['2024-03-01', '2024-03-02', '2024-03-01', '2024-03-01', '2024-03-03']
}
raw_logs = pd.DataFrame(data)
raw_logs
```

## What you should use
1. Pandas `.drop_duplicates()`
2. Pandas `.str.upper()` on the `cust_id` column.
3. Pandas `.fillna(0)` on the `amount` column.
4. `.groupby()` and `.agg()` to get the final result.


```python
# Optimized Solution

def clean_and_aggregate(df):
    # 1. Drop complete duplicates
    df_clean = df.drop_duplicates().copy()
    
    # 2. Standardize column cases
    df_clean['cust_id'] = df_clean['cust_id'].str.upper()
    
    # 3. Handle missing transaction amounts safely
    df_clean['amount'] = df_clean['amount'].fillna(0)
    
    # 4. Group and aggregate
    result = df_clean.groupby('cust_id', as_index=False).agg(total_amount=('amount', 'sum'))
    
    # Optional: Sort by total amount descending
    result = result.sort_values(by='total_amount', ascending=False).reset_index(drop=True)
    
    return result

final_df = clean_and_aggregate(raw_logs)
final_df
```
